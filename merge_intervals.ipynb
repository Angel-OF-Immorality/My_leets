```python
intervals = [[1,3],[2,6],[8,10],[15,18]]
```

```python
intervals.sort()
```

```python
intervals
```

```python
result = []
```

```python
curr_min = intervals[0][0]
curr_max = intervals[0][1]
for s,e in intervals[1:]:
    if s <= curr_max:
        curr_max = max(curr_max, e)
    else:
        result.append([curr_min, curr_max])
        curr_min = s
        curr_max = e
result.append([curr_min, curr_max])
return result
```
