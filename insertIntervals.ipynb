```python
intervals = [[1,2],[3,5],[6,7],[8,10],[12,16]]
newInterval = [4,8]
```

```python
result = []
new_s, new_e = newInterval
for i, (s,e) in enumerate(intervals):
    if e < new_s:
        result.append([s,e])
    elif s <= new_e:
        new_s = min(new_s, s)
        new_e = max(new_e, e)
    else:
         result.append([new_s, new_e])
         result.append([s,e])
         result.extend(intervals[i+1:])
         return result

result.append([new_s, new_e])
return result
```
