```python
arr = [0,3,7,2,5,8,4,6,0,1]
```

```python
#Using set
nums = set(arr)
result = 0
for i in nums:
    if i-1 not in nums:
        length = 1
        while (i + length) in nums:
            length += 1
        result = max(result, length)

print(result)
```

```python
#Using hashmap
map = {}
longest = 0

for i in arr:
    if i not in map:
        left = map.get(i-1, 0)
        right = map.get(i+1, 0)
        length = left + 1 + right
        map[i] = length
        #Update boundaries of current sequence because some element may be added at a boundary
        map[i - left] = length
        map[i + right] = length
        longest = max(longest, length)

print(longest)
```
