```python
pattern = "abba" 
s = "dog dog dog dog"
```

```python
words = s.split(' ')
words
```

```python
s_to_words = {}
words_to_s = {}
```

```python
for ch, word in zip(pattern, words):
    if ch in s_to_words and s_to_words[ch] != word:
        return False
    if word in words_to_s and words_to_s[word] != ch:
        return False

    s_to_words[ch] = word
    words_to_s[word] = ch

return True
```
