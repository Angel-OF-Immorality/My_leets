```python
nums = [0,1,2,4,5,7]
```

```python
result = []
```

```python
i = 0
while i < len(nums):
    a = nums[i]
    length = 1
    while i+length < len(nums) and a+length == nums[i+length]:
        length += 1
    if length == 1:
        result.append(str(a))
        i += 1
    else:
        result.append(str(a)+"->"+str(a+length-1))
        i += length
    
```
