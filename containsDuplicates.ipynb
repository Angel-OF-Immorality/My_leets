```python
nums = [1,2,3,1]
k = 3
```

```python
# nums[i] == nums[j] and abs(i-j) < k
```

```python
value_to_idx = {}
for i in range(len(nums)):
    if nums[i] in value_to_idx:
        if abs(value_to_idx[nums[i]] - i) <= k:
            print(True)
            break

    value_to_idx[nums[i]] = i
```
