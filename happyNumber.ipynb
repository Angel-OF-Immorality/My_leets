```python
n = 19
```

```python
num = n
set_of_squares = set()
sum_of_squares = 0

def get_square_array(num):
    digits = []
    while num > 0:
        digits.append((num%10)**2)
        num = num//10
    return digits
    

```

```python
get_square_array(19)
```

```python
while num not in set_of_squares:
    set_of_squares.add(sum_of_squares)
    digits = get_square_array(num)
    sum_of_squares = sum(digits)
    if sum_of_squares == 1:
        print(True)
        break
    num = sum_of_squares


```
