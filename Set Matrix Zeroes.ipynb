```python
matrix = [[0,1,2,0],[3,4,5,2],[1,3,1,5]]
```

```python
rows = len(matrix)
cols = len(matrix[0])
```

```python
rows, cols
```

```python
zero_pos = []
for x in range(rows):
    for y in range(cols): 
        if matrix[x][y] == 0:
            zero_pos.append([x,y])
```

```python
zero_pos
```

```python
for x,y in zero_pos:
    matrix[x] = [0]*cols
    for i in range(rows):
        matrix[i][y] = 0
```
