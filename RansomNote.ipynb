```python
ransomNote = "aabc"
magazine = "abc"
```

```python
from collections import Counter
counter1 = Counter(ransomNote)
counter1
```

```python
counter2 = Counter(magazine)
counter2
```

```python
for key in counter1.keys():
    if key not in counter2:
        return False

    else: 
        if counter1[key] > counter2[key]:
            return False
        
return True
```

```python
cnt =  (counter1 - counter2)
```
